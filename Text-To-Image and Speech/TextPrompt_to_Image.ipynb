```python
Diffusion model from Hugging Face to generate an image from a text prompt.

from IPython.display import display
from diffusers import AutoPipelineForText2Image
import torch

# Download/load an already-trained model. Another model we can use - stabilityai/stable-diffusion-xl-base-1.0
pipe = AutoPipelineForText2Image.from_pretrained("stabilityai/sdxl-turbo", torch_dtype=torch.float16, variant="fp16")
pipe.to("cuda")
prompt = "Prostate cancer show - adenocarcinoma, large cell carcinoma, squamous cell carcinoma, or normal (non-cancerous) lung tissue"
# Diffusion models normally start with noise and gradually transform it into an image
# SDXL Turbo is specifically designed for very fast image generation, 
# so it can produce reasonable results with very few inference steps.
image = pipe(prompt=prompt, num_inference_steps=4, guidance_scale=0.0).images[0]
display(image)
```
